```python
import numpy as np
import json
from datetime import datetime, timedelta
```

JSONB
Columnas base
product_weight_g
product_length_cm
product_height_cm
product_width_cm

```python
# TIPO AVANZADO: JSONB
# Basado en dimensiones reales del producto

products["product_specifications"] = products.apply(
    lambda row: {
        "weight_g": row["product_weight_g"],
        "length_cm": row["product_length_cm"],
        "height_cm": row["product_height_cm"],
        "width_cm": row["product_width_cm"]
    },
    axis=1
)

products[["product_id", "product_specifications"]].head()
```

RANGE (TSTZRANGE)
Columnas base
order_purchase_timestamp
order_estimated_delivery_date

(ORDERS)

```python
# TIPO AVANZADO: TSTZRANGE
# Ciclo real del pedido

orders["order_lifecycle"] = orders.apply(
    lambda row: f'["{row["order_purchase_timestamp"]}","{row["order_estimated_delivery_date"]}")',
    axis=1
)

orders[["order_id", "order_lifecycle"]].head()
```

TSVECTOR
Columnas base
review_comment_title
review_comment_message

```python
# TIPO AVANZADO: TSVECTOR
# Búsqueda textual

order_reviews["review_search"] = (
    order_reviews["review_comment_title"].fillna('') + ' ' +
    order_reviews["review_comment_message"].fillna('')
)

order_reviews[["review_id", "review_search"]].head()
```

POINT
Columnas base
geolocation_lat
geolocation_lng

```python
# TIPO AVANZADO: POINT
# Coordenadas reales

geolocation["geo_point"] = geolocation.apply(
    lambda row: (row["geolocation_lat"], row["geolocation_lng"]),
    axis=1
)

geolocation[["geolocation_zip_code_prefix", "geo_point"]].head()
```

HSTORE
Columnas base
payment_type
payment_installments
payment_value

```python
# TIPO AVANZADO: HSTORE
# Metadata de pago

order_payments["payment_metadata"] = order_payments.apply(
    lambda row: {
        "type": str(row["payment_type"]),
        "installments": str(row["payment_installments"]),
        "value": str(row["payment_value"])
    },
    axis=1
)

order_payments[["order_id", "payment_metadata"]].head()
```
